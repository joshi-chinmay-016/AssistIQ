# AssistIQ: Golden Evaluation Set Analysis

This notebook analyzes the **200-example human-verified golden evaluation set** for **AssistIQ** (SpotifyCares Twitter Customer Support).

### Overview
- File: `dataset/golden_set.csv`
- Total examples: 200 (100% human-verified)
- Columns: `tweet_id`, `text`, `candidate_intent`, `human_verified_intent`, `review_status`

### 1. Intent Taxonomy (11 Classes)
1. `playback_and_app_issues` - Audio errors, streaming glitches, app crashes, freezes, device connectivity (Bluetooth/HiFi).
2. `search_and_discovery` - Search bar, recommendation algorithms, Discover Weekly, Daily Mix, finding albums.
3. `account_and_login` - Credentials, password resets, CSRF errors, hacked accounts, 2FA, abroad/region locks.
4. `billing_and_payment` - Charges, double renewals, payment methods (credit cards, gift cards, Dragonpay, Oxxo).
5. `premium_and_subscription` - Upgrades, Family plan members, Student discount verification (SheerID/UNiDAYS), trials.
6. `music_availability` - Missing songs/albums, regional licensing restrictions, artist distributor delays (CDBaby).
7. `playlist_and_library` - Disappeared playlists, wiped saved music, cleared offline downloads, playlist editing.
8. `feature_requests` - Sleep timer, clean lyrics toggle, visualizer, UI settings, incognito mode, hold music.
9. `content_metadata` - Wrong audio track, incorrect artist credit, missing album release date, lyrics typos.
10. `ads_and_privacy` - Popups (e.g. Waze popups), ads on premium accounts, privacy policies, data tracking.
11. `other_non_actionable` - Thanks, banter, emojis, screenshots without context, version dumps without active issue.

```python
import os
import pandas as pd

golden_path = "../dataset/golden_set.csv"
if not os.path.exists(golden_path):
    golden_path = "dataset/golden_set.csv"

df = pd.read_csv(golden_path)
print("Golden Set Shape:", df.shape)
print("\nReview Status Breakdown:")
print(df['review_status'].value_counts())
print("\nMissing values:")
print(df.isna().sum())
df.head()
```

### 2. Candidate vs Human-Verified Intent Analysis
Inspect rows where the final human-verified intent corrected or diverged from the initial candidate intent.

```python
disagreements = df[df['candidate_intent'] != df['human_verified_intent']]
agreement_rate = (len(df) - len(disagreements)) / len(df) * 100
print(f"Agreement Rate: {agreement_rate:.1f}% ({len(df) - len(disagreements)} / {len(df)})")
print(f"Total Corrections / Disagreements: {len(disagreements)}\n")

for idx, row in disagreements.iterrows():
    print(f"[Tweet ID: {row.tweet_id}]")
    print(f"  Text:            {row.text}")
    print(f"  Candidate:       {row.candidate_intent}")
    print(f"  Human Verified:  {row.human_verified_intent}")
    print("-" * 80)
```

### 3. Final Golden Evaluation Set Distribution
Inspect the distribution of human-verified intents across all 11 taxonomy classes.

```python
dist_df = pd.DataFrame({
    'Candidate Intent': df['candidate_intent'].value_counts(),
    'Human-Verified Ground Truth': df['human_verified_intent'].value_counts()
}).fillna(0).astype(int)
dist_df.sort_values(by='Human-Verified Ground Truth', ascending=False)
```
